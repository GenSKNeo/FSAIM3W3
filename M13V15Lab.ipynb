# Sample Code from Lecture Notes
Note: The data is loaded from the numpy array values. Numpy is a fundamental package for scientific computing in Python, providing support for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays efficiently.

```python
# Importing necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Creating the input data for the model
xvals = np.array([-1, 0, 1, 2, 3, 4, 5, 6, 7]).reshape((-1, 1))
yvals = np.array([-4, 1, 6, 11, 16, 21, 26, 31, 36]).reshape((-1, 1))

# Initializing the Linear Regression model
model = LinearRegression()

# Fitting the model with the input data
model.fit(xvals, yvals)
# Creating new data to make predictions
newx = np.array([8, 9, 10]).reshape((-1, 1))

# Making predictions with the new data
y_pred = model.predict(newx)

# Printing the predicted values
print('Predicted y-values:', y_pred, sep='\n')
```

# **Load Data from CSV**
To modify the previous example to load data from a CSV file for linear regression, you'll need to use the pandas library, which provides excellent support for reading and manipulating data files.
Pandas is a library for data manipulation and analysis, offering data structures and operations for manipulating numerical tables and time series. First you need to upload the sample data.csv file that is given to you to a folder in your google drive. Then you need to get the path of this file and replace it in the below code.
Below is how you can adjust the code to incorporate data loading from a CSV file:



```python
# Importing necessary libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

path = '/content/drive/MyDrive/data.csv'

# Load data from a CSV file
# Ensure your CSV file is in the format of two columns: x values and y values without headers
data = pd.read_csv(path, header=None, names=['x', 'y'])

# Reshape data to fit the model: '-1' lets numpy determine the number of rows based on original data and number of columns
xvals = data['x'].values.reshape((-1, 1))  # Reshape x into a 2D array for sklearn
yvals = data['y'].values.reshape((-1, 1))  # Reshape y into a 2D array for sklearn

# Initialize the Linear Regression model
model = LinearRegression()

# Train the model with the dataset
model.fit(xvals, yvals)

# New data for prediction
newx = np.array([8, 9, 10]).reshape((-1, 1))  # New x values for which we want predictions

# Make predictions based on the new x values
y_pred = model.predict(newx)

# Print out the predicted y-values
print('Predicted y-values:', y_pred, sep='\n')
print(data)

# Output will show predicted y-values for the input array [8, 9, 10]
```
